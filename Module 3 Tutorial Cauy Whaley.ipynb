## Exercise 1 — Sort an Array of 0s, 1s, and 2s

The three-pointer approach groups 0s at the beginning, 1s in the middle, and 2s at the end. When swapping a 2 with the high pointer, the middle pointer stays in place so the swapped-in value can be examined.

```python
class Solution:
    def sort012(self, arr):
        low = 0
        mid = 0
        high = len(arr) - 1

        while mid <= high:
            if arr[mid] == 0:
                arr[low], arr[mid] = arr[mid], arr[low]
                low += 1
                mid += 1
            elif arr[mid] == 1:
                mid += 1
            else:
                arr[mid], arr[high] = arr[high], arr[mid]
                high -= 1

        return arr
```

```python
sorter = Solution()
for numbers in ([0, 1, 2, 0, 1, 2], [2, 0, 1, 2, 0], [1, 1, 1], []):
    actual = sorter.sort012(list(numbers))
    expected = sorted(numbers)
    print(f"Input: {numbers} -> Output: {actual}")
    assert actual == expected
print("All sorting tests passed.")
```

## Exercise 2 — Binary Search (First Occurrence)

The input array must already be sorted. The algorithm narrows the search range by half on each iteration. When it finds the target, it records the index and continues searching left to find the first occurrence. It returns `-1` if the target is absent.

```python
class Solution:
    def firstSearch(self, arr, k):
        low = 0
        high = len(arr) - 1
        result = -1

        while low <= high:
            mid = (low + high) // 2
            if arr[mid] == k:
                result = mid
                high = mid - 1
            elif arr[mid] < k:
                low = mid + 1
            else:
                high = mid - 1

        return result
```

```python
searcher = Solution()
examples = [([1, 2, 3, 4, 5], 4, 3), ([1, 1, 1, 2, 3], 1, 0), ([10, 20, 30, 40], 50, -1), ([2, 4, 6, 8, 10], 6, 2), ([], 2, -1)]
for numbers, target, expected in examples:
    actual = searcher.firstSearch(numbers, target)
    print(f"Array: {numbers}, target: {target} -> Index: {actual}")
    assert actual == expected
print("All binary search tests passed.")
```
